The goal of this project is to help a telecom business identify customers who are likely to churn and understand different customer segments.

## 9. Check for missing values

Missing values can affect analysis, feature engineering, and model training.

First, we check the number of actual null values in every column.

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

### Make the result easier to read

Instead of looking at a long Series, we create a small table containing only columns where null values exist.

If this table is empty, there are no explicit null values.

```python
missing_value_summary = pd.DataFrame({
    "Column": missing_values.index,
    "Missing_Values": missing_values.values
})

missing_value_summary = missing_value_summary[
    missing_value_summary["Missing_Values"] > 0
]

if missing_value_summary.empty:
    print("No explicit missing values were found.")
else:
    display(missing_value_summary)
```

### Finding

The raw dataset reports **0 explicit null values**.

However, this does not completely prove that the dataset has no missing information.

A text column can contain an empty string or whitespace instead of a true null value.

So we perform one more check.

## 10. Check for blank and whitespace values

This check is especially important for text columns.

For example:

`""`

and

`"   "`

are not necessarily detected by `isnull()`.

We therefore check whether any text column contains values that become empty after removing surrounding whitespace.

```python
text_columns = df.select_dtypes(include="object").columns

blank_value_counts = {}

for column in text_columns:
    blank_count = df[column].fillna("").astype(str).str.strip().eq("").sum()
    blank_value_counts[column] = blank_count

blank_value_summary = pd.Series(
    blank_value_counts,
    name="Blank_or_Whitespace_Values"
)

blank_value_summary = blank_value_summary[
    blank_value_summary > 0
]

if blank_value_summary.empty:
    print("No blank or whitespace-only values were found in text columns.")
else:
    display(blank_value_summary)
```

## 11. Check for duplicate rows

Duplicate records can distort customer counts and downstream analysis.

We first check for exact duplicate rows, where every value across the row is identical.

```python
duplicate_rows = df.duplicated().sum()

print(f"Number of exact duplicate rows: {duplicate_rows:,}")
```

```python
if duplicate_rows == 0:
    print("No exact duplicate rows were found.")
else:
    print("Duplicate rows were found and need investigation before modelling.")
```

### Finding

The dataset contains **0 exact duplicate rows**.

Therefore, there is no evidence of exact row duplication at this stage.

We will still separately check whether the customer identifier is unique.

## 12. Validate the customer identifier

`customerID` is an identifier.

An identifier is useful for tracing a prediction back to a customer, but it should generally not be used as a predictive feature because the ID itself does not represent customer behaviour.

Before making that decision, we check whether each customer ID is unique.

```python
customer_id_column = "customerID"

number_of_unique_customers = df[customer_id_column].nunique()
number_of_customer_records = len(df)

print(f"Total customer records: {number_of_customer_records:,}")
print(f"Unique customer IDs: {number_of_unique_customers:,}")
```

```python
duplicate_customer_ids = df[customer_id_column].duplicated().sum()

print(f"Duplicate customer IDs: {duplicate_customer_ids:,}")
```

```python
missing_customer_ids = df[customer_id_column].isnull().sum()

print(f"Missing customer IDs: {missing_customer_ids:,}")
```

### Decision

If every row has one unique customer ID and there are no missing IDs, we can use `customerID` for traceability while excluding it from the predictive feature set later.

We are **not dropping it from the raw dataset**. That decision belongs to the modelling pipeline.

## 13. Validate expected numeric columns

A column should have a data type that makes sense for its business meaning.

For this dataset, we expect these fields to be numerical:

- `SeniorCitizen`
- `tenure`
- `MonthlyCharges`
- `TotalCharges`

Let's check their actual types.

```python
expected_numeric_columns = [
    "SeniorCitizen",
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

for column in expected_numeric_columns:
    print(f"{column:20} -> {df[column].dtype}")
```

### Finding

`TotalCharges` is stored as a string rather than a numeric type.

This does not automatically mean that the values are wrong. It means we need to investigate the actual contents before cleaning the column.

## 14. Investigate `TotalCharges`

`TotalCharges` should represent the total amount charged to a customer.

We will test which values can be converted to numeric values.

We use `errors="coerce"` only for this validation test. We are **not changing the original DataFrame** here.

```python
total_charges_as_numeric = pd.to_numeric(
    df["TotalCharges"].astype(str).str.strip(),
    errors="coerce"
)

conversion_failures = total_charges_as_numeric.isnull().sum()

print(
    f"Number of TotalCharges values that cannot be converted to numeric: "
    f"{conversion_failures:,}"
)
```

If conversion failures exist, we need to see the original values responsible for them.

```python
if conversion_failures > 0:

    problematic_total_charges = df.loc[
        total_charges_as_numeric.isnull(),
        "TotalCharges"
    ]

    print("Problematic raw TotalCharges values:")
    display(problematic_total_charges.value_counts(dropna=False))

else:
    print("All TotalCharges values can be converted to numeric.")
```

### Validation decision

We have now identified the problem without modifying the raw dataset.

The actual cleaning rule for `TotalCharges` will be decided in the **Data Cleaning** stage, where we can also compare the affected records with `tenure` and other relevant fields.

This keeps validation and cleaning separate.

## 15. Validate categorical columns

Categorical variables should contain consistent values.

For example, if a column is expected to contain:

`Yes`, `No`

we should not silently have values such as:

`yes`, `YES`, `Y`, `No `

We inspect the unique values and their frequencies before deciding whether any standardization is required.

```python
categorical_columns = df.select_dtypes(include="object").columns

print("Categorical columns:")
for column in categorical_columns:
    print(f"- {column}")
```

### Inspect category frequencies

We will inspect the categories column by column.

This is intentionally done as a validation step. We are not changing category labels yet.

```python
for column in categorical_columns:

    print()
    print("=" * 70)
    print(f"Column: {column}")
    print("=" * 70)

    value_counts = df[column].value_counts(dropna=False)

    display(value_counts.to_frame(name="Count"))
```

### What we are looking for

During this review, pay attention to:

- unexpected spelling variations
- leading or trailing spaces
- unexpected null-like values
- categories that do not make business sense
- inconsistent representations of the same concept

The goal is not to remove categories just because they are uncommon.

## 16. Basic numerical sanity checks

Now we validate whether numerical values fall into sensible ranges.

This is different from outlier analysis.

For example, a very high monthly charge might be a legitimate customer. But a negative monthly charge would require investigation.

We therefore check obvious business constraints first.

```python
print("Summary of numerical variables:")
df[["SeniorCitizen", "tenure", "MonthlyCharges"]].describe()
```

### Check `SeniorCitizen`

This dataset uses:

- `0` = not a senior citizen
- `1` = senior citizen

Therefore, values outside 0 and 1 would indicate a data-quality issue.

```python
invalid_senior_citizen_values = df[
    ~df["SeniorCitizen"].isin([0, 1])
]

print(
    f"Number of records with invalid SeniorCitizen values: "
    f"{len(invalid_senior_citizen_values):,}"
)
```

### Check `tenure`

Customer tenure represents the number of months the customer has stayed with the company.

Negative tenure would not make business sense.

```python
negative_tenure_records = df[df["tenure"] < 0]

print(
    f"Number of records with negative tenure: "
    f"{len(negative_tenure_records):,}"
)
```

### Check `MonthlyCharges`

Monthly charges should not be negative.

A value of zero may or may not require business investigation, but a negative charge is clearly suspicious.

```python
negative_monthly_charge_records = df[
    df["MonthlyCharges"] < 0
]

print(
    f"Number of records with negative MonthlyCharges: "
    f"{len(negative_monthly_charge_records):,}"
)
```

## 17. Validate the target variable

The target variable for our classification problem is:

`Churn`

Before modelling, we need to confirm:

1. Which labels exist?
2. Are there unexpected labels?
3. How many customers belong to each class?

```python
target_column = "Churn"

print("Unique target values:")
print(df[target_column].unique())
```

```python
target_counts = df[target_column].value_counts(dropna=False)

print("Number of customers in each churn class:")
print(target_counts)
```

```python
target_percentages = (
    df[target_column]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
)

print("Percentage of customers in each churn class:")
print(target_percentages)
```

### Finding

The dataset contains:

- **5,174 customers who did not churn**
- **1,869 customers who churned**

That means approximately:

- **73.46% No Churn**
- **26.54% Churn**

This is not a reason to remove or artificially rebalance the dataset at this stage.

However, it tells us something important for modelling:

> Accuracy alone will not be enough to evaluate our churn model.

Later we will consider metrics such as recall, precision, F1-score and probability/threshold behaviour.

## 18. Check for obvious target problems

We expect the target to contain only:

- `Yes`
- `No`

Let's explicitly validate that assumption.

```python
expected_churn_values = {"Yes", "No"}

actual_churn_values = set(
    df[target_column].dropna().unique()
)

unexpected_churn_values = actual_churn_values - expected_churn_values

if len(unexpected_churn_values) == 0:
    print("Target validation passed: only Yes and No labels are present.")
else:
    print("Unexpected target labels were found:")
    print(unexpected_churn_values)
```

## 19. Basic leakage screening

Data leakage happens when a model receives information that would not realistically be available when the prediction is made.

At this stage we only perform an initial screening.

We will make final feature decisions later after understanding the business problem and the prediction timing.

```python
print("Columns that require timing/business review before modelling:")
print()

columns_to_review = [
    "tenure",
    "TotalCharges",
    "Contract",
    "PaymentMethod"
]

for column in columns_to_review:
    print(f"- {column}")
```

### Important note

A column is not automatically leakage just because it is strongly related to churn.

The real question is:

> **Would this information be available at the exact moment when the business wants to make the churn prediction?**

We will revisit this during feature engineering and modelling.
